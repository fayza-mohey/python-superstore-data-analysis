data cleaning

```python
orders = orders.drop_duplicates()
orders
```

Outliers

```python
plt.Figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.boxplot(y=orders["Sales"])
plt.title("Sales - before capping outliers")

plt.subplot(1, 2, 2)
sns.boxplot(y=orders["Profit"])
plt.title("Profit - before capping outliers")

plt.tight_layout()
plt.show()
```

Feature engineering

```python
# Profit Margin = how much profit per dollar of sales.
orders["Profit Margin"] = orders["Profit"] / orders["Sales"]

# Shipping Duration = how many days for order shipping
orders["Order Date"] = pd.to_datetime(orders["Order Date"])
orders["Ship Date"] = pd.to_datetime(orders["Ship Date"])
orders["Shipping Duration"] = (orders["Ship Date"] - orders["Order Date"]).dt.days

# Sales Performance Category = to split sales into 4 groups (Low, Medium, High, Top)
orders["Sales Performance Category"] = pd.qcut(orders["Sales"], q=4, labels=["Low", "Medium", "High", "Top"])

orders["Order Year"] = orders["Order Date"].dt.year
orders["Order Month"] = orders["Order Date"].dt.month

orders[["Sales", "Profit", "Profit Margin", "Shipping Duration", "Sales Performance Category"]].head()
```

A simple class to keep the analysis organized

Just to show basic OOP - a class that stores the cleaned data and has a couple of helper methods.

```python
class SalesAnalyzer:

    def __init__(self, df):
        self.df = df

    def sales_by(self, column):
        # total sales and profit grouped by a column
        result = self.df.groupby(column)[["Sales", "Profit"]].sum()
        result = result.sort_values("Sales", ascending=False)
        return result

    def top_customers(self, n=5):
        return self.df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(n)


analyzer = SalesAnalyzer(orders)
```

```python
print("Sales and Profit by Category:")
analyzer.sales_by("Category")
```

```python
print("Sales and Profit by Region:")
analyzer.sales_by("Region")
```

```python
print("Sales and Profit by Segment:")
analyzer.sales_by("Segment")
```

```python
print("Top 5 customers by sales:")
analyzer.top_customers()
```

Trend analysis over time

```python
orders["Order Year-Month"] = orders["Order Date"].dt.to_period("M").astype(str)

monthly_sales = orders.groupby("Order Year-Month")["Sales"].sum().reset_index()
monthly_sales.head()
```

```python
plt.figure(figsize=(12, 5))
plt.plot(monthly_sales["Order Year-Month"], monthly_sales["Sales"], marker="o")
plt.xticks(rotation=90)
plt.title("Monthly Sales Over Time")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

# KPI summary

```python
total_sales = orders["Sales"].sum()
total_profit = orders["Profit"].sum()
profit_margin = total_profit / total_sales * 100
total_orders = orders["Order ID"].nunique()
avg_order_value = orders.groupby("Order ID")["Sales"].sum().mean()
avg_shipping_days = orders["Shipping Duration"].mean()
return_rate = returns["Order ID"].nunique() / total_orders * 100

print("---- KPI SUMMARY ----")
print("Total Sales: $", round(total_sales, 2))
print("Total Profit: $", round(total_profit, 2))
print("Profit Margin: ", round(profit_margin, 2), "%")
print("Total Orders: ", total_orders)
print("Average Order Value: $", round(avg_order_value, 2))
print("Average Shipping Duration: ", round(avg_shipping_days, 2), "days")
print("Return Rate: ", round(return_rate, 2), "%")
```
